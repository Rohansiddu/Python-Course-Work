```python
import pandas as pd

prices=[2599,15999,4999,1999,2999]
products=["wireless Earbuds","smartphone","laptop","smartwatch","Bluetooth speaker"]
product_prices=pd.Series(prices,index=products)
print(product_prices)
```

```python
print("mean:" , product_prices.mean())
print("max:" , product_prices.max())
print("min:" , product_prices.min())
```

```python
print("head (First 3 Elements): \n", product_prices.head(2))
print("tail (Last 3 Elements): \n", product_prices.tail(1))
```

Use Apply for Complex things using Lambda:

```python
print("Apply (adding 18% GST): \n", product_prices.apply(lambda x: f'₹{x+(x * 1.18)}'))
```

Use map for simple structure:

```python
print("Map (formatting as currency): \n", product_prices.map(lambda x: f'₹{x}.00'))
```

```python
print("sort by values: \n", product_prices.sort_values())
```

```python
print("sort by Index: \n", product_prices.sort_index())
```

```python
print(product_prices.sort_index(ascending=False))
```

```python
print("count of values: \n", product_prices.value_counts())
```

```python
data={
    "Product":["wireless Earbuds", "Smartphome", "Laptop", "Smartwatch", "Bluetooth speaker"],
    "Brand": ["SoundMax", "TechNova","ByteCore","TimeTrack","EchoBoom"],
    "Price": [2999,15999,52999,4999,5678],
    "Stock": [50,30,20,40,60],
    "BestSeller": [True,False,False,True, True]
}
df = pd.DataFrame(data)
print(df)
```

```python
print('shape of the dataframe: ',df.shape) #shape of the dataframe
print('colums in Dataframe: ',df.columns) #colums in Dataframe
print('index of the dataframe: ',df.index) #index of the dataframe
```

```python
print('\t dataFrame info: \n')
print(df.info()) #dataFrame info
```

```python
print('\t Statistical Summary: \n')
print(df.describe()) #Statistical Summary
```

```python
print('\tHead (First 3 rows): \n ', df.head(3)) #Head (First 3 rows)
print('\n\n\ttail (last 3 rows): \n'), df.tail(3)
```

```python
print(df,end='\n\n')
print(df.loc[1,'Product'])
print(df.loc[4,'Price'])
print(df.loc[4, 'Stock'])

print(df.iloc[4,0])
print(df.iloc[2,1])
print(df.iloc[3,3])
```

```python
print('using iloc (selecrting 4th row): \n',df.iloc[4])
```

```python
print(df)
print('************Bestseller using loc**************\n',df.loc[df['BestSeller']]==True)
print('************Stock Greater less than 40 using iloc**************\n',df.loc[(df['Stock']<40)])
print('************Price lessthen 10000 using iloc**************\n',df.loc[(df['Price']<10000)])
```

```python
print("before df:",df)
df_dropped = df.drop(columns=["Stock"])
print("After dropping 'stock' Column: \n", df_dropped)
print("after df:", df)
```

```python
df_renamed=df.rename(columns={"Product":"Product name"}) #inplace=True is for chane in the dataframe
print(df_renamed)
```

```python
df_grouped = df.groupby("Brand").agg({"Price": "sum", "Stock": "sum"})
print(df_grouped)
```

```python
grouped = df.groupby("Brand")["Price"].mean()
print(grouped)
```
